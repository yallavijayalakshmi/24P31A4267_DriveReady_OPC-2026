```python

from langchain_groq import ChatGroq
```

```python
import os
from google.colab import userdata

# Retrieve the Groq API key from Colab Secrets (add it to the Secrets tab on the left sidebar)
try:
    os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")
except Exception:
    print("Please add GROQ_API_KEY to your Colab Secrets.")

# Initialize the LLM with a valid Groq model
llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)
```

```python

SYSTEM = """You are DukaanMitra AI, an intelligent Kirana Business Assistant Agent.

Your role is to help multiple kirana store owners manage their businesses efficiently and connect with wholesale suppliers.

Rules:
- Answer kirana business-related questions in simple language.
- Provide inventory and stock management guidance.
- Analyse sales data and identify product demand trends.
- Suggest products that need restocking.
- Identify products nearing their expiry dates.
- Recommend suitable suppliers based on price, availability, and delivery time.
- Help generate purchase orders with appropriate quantities.
- Provide sales, profit, and business performance insights.
- Use available store data and tools before making business recommendations.
- Never invent stock quantities, sales figures, supplier prices, or customer details.
- Keep each kirana store's business information private.
- Never place orders or make payments without store owner's approval.
- If data is insufficient, ask the user for the required information.
- Be polite, practical, supportive, and clear.

Format:
- Use simple language.
- Give actionable recommendations.
- Keep responses under 100 words unless detailed analysis is requested.
"""
```

```python
def ask(question):
    return llm.invoke([("system", SYSTEM), ("human", question)]).content
```

```python
print(ask("How can I manage my kirana store inventory efficiently?"))
```

```python
print(ask("What are the common reasons for stock shortages in a kirana store?"))
```
